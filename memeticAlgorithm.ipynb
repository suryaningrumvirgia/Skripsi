## Clase City:

Clase City didefinisikan dengan metode dan atribut berikut:

Atribut:
+ x: variabel yang menyimpan koordinat x titik.
+ y: variabel yang menyimpan koordinat y titik.
+ delivery_time: waktu delay pada kota yang disimulasikan, default 0.

Metode:
+ distance(): menerima objek tipe City, menghitung jarak antara titik saat ini dan titik yang diterima, kemudian mengembalikan jarak tersebut.
+ delivery_duration(): menghitung waktu pengiriman antar dua kota berdasarkan jarak dan rata-rata kecepatan.



```python
class City:
    def __init__(self, x, y, delivery_time=0):
        self.x = x
        self.y = y
        self.delivery_time = delivery_time  # Waktu delay untuk setiap kota.

    def distance(self, city):
        xDis = abs(self.x - city.x)
        yDis = abs(self.y - city.y)
        distance = np.sqrt((xDis ** 2) + (yDis ** 2))
        return distance

    def delivery_duration(self, city, average_speed=50):
        # Hitung waktu perjalanan antara dua kota berdasarkan kecepatan rata-rata
        travel_time = self.distance(city) / average_speed
        return travel_time

    def __repr__(self):
        return f"({self.x},{self.y},{self.delivery_time})"
```

## Clase Fitness

Clase fitness didefinisikan dengan metode dan atribut berikut:

Atribut:
+ rute: parameter input dalam konstruktor, menerima rute yang dimiliki oleh individu.
+ total_time: variabel yang menyimpan total waktu pengiriman rute, termasuk delay di kota.
+ fitness: variabel yang menyimpan kinerja individu, diukur berdasarkan kebalikan dari total waktu.

Metode:
+ routeTime(): menghitung dan mengembalikan total waktu pengiriman (termasuk delay setiap kota).
+ routeFitness(): menghitung kebugaran individu, yang ditetapkan sebagai 1/total waktu, mengembalikan dan menyimpan nilai tersebut ke variabel fitness.


```python
class Fitness:
    def __init__(self, route):
        self.route = route
        self.total_time = 0  # Dengan waktu termasuk delay kota
        self.fitness = 0.0

    def routeTime(self, average_speed=50):
        if self.total_time == 0:
            total_time = 0
            for i in range(0, len(self.route)):
                fromCity = self.route[i]
                toCity = self.route[i + 1] if i + 1 < len(self.route) else self.route[0]
                travel_time = fromCity.delivery_duration(toCity, average_speed)
                total_time += travel_time + fromCity.delivery_time  # Tambahkan waktu delay kota
            self.total_time = total_time
        return self.total_time

    def routeFitness(self):
        if self.fitness == 0:
            self.fitness = 1 / float(self.routeTime())  # Semakin kecil waktu, semakin besar nilai fitness
        return self.fitness
```
